```python
# Cities list
cities = ["Louisville", "New York", "Lihue", "Kapaa", "Morehead"]

# Empty list to store summaries
city_summaries = []

# Function to get city weather summary
def get_city_summary(city, api_key):

    # Current weather
    current_url = f"{baseurl}/current.json?key={api_key}&q={city}"
    current_response = requests.get(current_url)
    current_data = current_response.json()

    # Forecast weather
    forecast_url = f"{baseurl}/forecast.json?key={api_key}&q={city}&days=3"
    forecast_response = requests.get(forecast_url)
    forecast_data = forecast_response.json()

    # Astronomy data
    today = datetime.now().strftime("%Y-%m-%d")

    astronomy_url = (
        f"{baseurl}/astronomy.json?key={api_key}&q={city}&dt={today}"
    )

    astronomy_response = requests.get(astronomy_url)
    astronomy_data = astronomy_response.json()

    # Build city summary dictionary
    city_summary = {
        "City": city,
        "Country": current_data['location']['country'],
        "Temperature_F": current_data['current']['temp_f'],
        "Temperature_C": current_data['current']['temp_c'],
        "Condition": current_data['current']['condition']['text'],
        "Humidity": current_data['current']['humidity'],
        "Wind_MPH": current_data['current']['wind_mph'],
        "Sunrise": astronomy_data['astronomy']['astro']['sunrise'],
        "Sunset": astronomy_data['astronomy']['astro']['sunset'],
        "Forecast_Day1_AvgTemp_F":
            forecast_data['forecast']['forecastday'][0]['day']['avgtemp_f'],
        "Forecast_Day2_AvgTemp_F":
            forecast_data['forecast']['forecastday'][1]['day']['avgtemp_f'],
        "Forecast_Day3_AvgTemp_F":
            forecast_data['forecast']['forecastday'][2]['day']['avgtemp_f']
    }

    return city_summary


# Loop through all cities
for city in cities:

    
    summary = get_city_summary(city, api_key)
    city_summaries.append(summary)

    
# Create DataFrame
weather_df = pd.DataFrame(city_summaries)
```

```python
# Save to CSV
weather_df.to_csv("city_weather_summary.csv", index=False)
```
